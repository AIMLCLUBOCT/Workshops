## 1. Creating and Inspecting NumPy Arrays
NumPy arrays (`ndarray`) are homogeneous, contiguous blocks of memory optimized in C.

```python
# 1D Vector
vector = np.array([1.0, 2.0, 3.0, 4.0])
print('Vector shape:', vector.shape)

# 2D Matrix (e.g. Weights / Feature Data)
matrix = np.array([[1, 2, 3], [4, 5, 6]], dtype=np.float32)
print('Matrix:\n', matrix)
print('Shape:', matrix.shape, '| Dtype:', matrix.dtype)
```

## 2. Speed Comparison: Vectorization vs. Native Python Loops
Let us benchmark adding 1,000,000 numbers in Python lists vs. NumPy vectorization.

```python
n = 1_000_000
py_list = list(range(n))
np_arr = np.arange(n)

# Python loop
t0 = time.time()
res_py = [x + 2 for x in py_list]
t_py = time.time() - t0

# NumPy vectorized addition
t0 = time.time()
res_np = np_arr + 2
t_np = time.time() - t0

print(f'Native Python loop: {t_py:.4f}s')
print(f'NumPy vectorization: {t_np:.4f}s')
print(f'NumPy is {t_py / max(t_np, 1e-6):.1f}x faster!')
```

## 3. Hands-On Exercise: Vectorized Euclidean Distance
Calculate Euclidean distance $d(p, q) = \sqrt{\sum (p_i - q_i)^2}$ without using any loops.

```python
def euclidean_distance(v1: np.ndarray, v2: np.ndarray) -> float:
    diff = v1 - v2
    return float(np.sqrt(np.sum(diff ** 2)))

p1 = np.array([2.0, 3.0, 5.0])
p2 = np.array([5.0, 7.0, 9.0])
dist = euclidean_distance(p1, p2)
print(f'Distance between p1 and p2: {dist:.4f}')
```
