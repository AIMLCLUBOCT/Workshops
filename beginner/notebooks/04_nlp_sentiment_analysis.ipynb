## 1. Setup & Environment Dependencies
Let's import the necessary libraries (`scikit-learn`, `pandas`, `numpy`, and `matplotlib`).

```python
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay

print('Libraries imported successfully!')
```

## 2. Load Sample Sentiment Dataset
We use a dataset containing balanced positive and negative reviews.

```python
data = [
    # Positive reviews (1)
    ('This AI workshop was extremely informative and well-structured!', 1),
    ('The practical coding demonstrations helped clarify the core concepts.', 1),
    ('Brilliant hands-on lab! I learned so much about neural networks.', 1),
    ('The mentors were patient, encouraging, and answered all questions.', 1),
    ('Loved the step-by-step approach to machine learning fundamentals.', 1),
    ('The project starter code was clean, reproducible, and easy to run.', 1),
    ('Great session! Really looking forward to the next hackathon.', 1),
    ('Outstanding explanations of linear algebra and loss functions.', 1),
    ('I highly recommend this club workshop to all engineering students.', 1),
    ('Awesome experience building my first sentiment classification model!', 1),
    # Negative reviews (0)
    ('The pacing was way too fast and hard to follow for beginners.', 0),
    ('Disappointed with the sound quality in the seminar hall.', 0),
    ('Too much theory on slides and not enough interactive coding time.', 0),
    ('The prerequisites were not clearly stated and I felt completely lost.', 0),
    ('The example code had errors and failed to run on my local machine.', 0),
    ('The workshop started late and ended before covering all topics.', 0),
    ('Unhelpful explanations, it felt rushed and disorganized.', 0),
    ('The presentation was boring and repetitive.', 0),
    ('Could not get the environment setup properly in time.', 0),
    ('Poor communication regarding the schedule and lab location.', 0)
]

df = pd.DataFrame(data, columns=['review', 'sentiment'])
print(f'Total reviews: {len(df)}')
df.head()
```

## 3. Text Preprocessing Function
Raw text data contains punctuation, irregular casing, and noise. We normalize text into clean tokens.

```python
def clean_text(text: str) -> str:
    text = text.lower() # Lowercase
    text = re.sub(r'[^a-zA-Z0-9\s]', ' ', text) # Strip punctuation
    text = re.sub(r'\s+', ' ', text).strip() # Normalize whitespace
    return text

df['cleaned_review'] = df['review'].apply(clean_text)
df[['review', 'cleaned_review']].head()
```

## 4. Train / Test Split & TF-IDF Vectorization
We separate training data (75%) from holdout test data (25%) and vectorize using unigrams and bigrams.

```python
X_train, X_test, y_train, y_test = train_test_split(
    df['cleaned_review'], df['sentiment'], test_size=0.25, random_state=42, stratify=df['sentiment']
)

vectorizer = TfidfVectorizer(ngram_range=(1, 2), stop_words='english')
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

print(f'Training shape: {X_train_vec.shape}')
print(f'Testing shape:  {X_test_vec.shape}')
```

## 5. Model Training & Evaluation
We fit a **Multinomial Naive Bayes** classifier and evaluate precision, recall, and the confusion matrix.

```python
model = MultinomialNB(alpha=1.0)
model.fit(X_train_vec, y_train)

preds = model.predict(X_test_vec)
print('Classification Report:')
print(classification_report(y_test, preds, target_names=['Negative', 'Positive']))

# Display Confusion Matrix
cm = confusion_matrix(y_test, preds)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Negative', 'Positive'])
disp.plot(cmap='Blues')
plt.title('Sentiment Confusion Matrix')
plt.show()
```

## 6. Interactive Live Inference
Test the classifier with your own review sentences!

```python
def predict_sentiment(text: str):
    cleaned = clean_text(text)
    vec = vectorizer.transform([cleaned])
    pred = model.predict(vec)[0]
    proba = model.predict_proba(vec)[0]
    sentiment = '😊 Positive' if pred == 1 else '😞 Negative'
    confidence = proba[pred] * 100
    print(f'Input:      "{text}"')
    print(f'Sentiment:  {sentiment} (Confidence: {confidence:.2f}%)\n')

predict_sentiment('The instructors explained every concept with fantastic examples!')
predict_sentiment('I had a terrible experience with the code errors and confusing slides.')
```
