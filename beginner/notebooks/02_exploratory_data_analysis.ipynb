```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
print('EDA libraries imported successfully!')
```

## 1. Creating and Inspecting a Sample Dataset

```python
np.random.seed(42)
n_samples = 400

df = pd.DataFrame({
    'Study_Hours': np.random.uniform(1.0, 10.0, n_samples),
    'Attendance_Rate': np.random.uniform(50.0, 100.0, n_samples),
    'Branch': np.random.choice(['CSE', 'AIML', 'IT', 'ECE'], n_samples),
})
# Exam score correlates with study hours and attendance
df['Exam_Score'] = np.clip(10 + 5.5 * df['Study_Hours'] + 0.35 * df['Attendance_Rate'] + np.random.normal(0, 5, n_samples), 0, 100)
df['Passed'] = (df['Exam_Score'] >= 40).astype(int)

print('Shape:', df.shape)
df.head()
```

## 2. Summary Statistics & Missing Values Check

```python
print('Missing Values:\n', df.isnull().sum())
display(df.describe())
```

## 3. Data Visualization: Distribution & Correlation Heatmap

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Scatter plot with regression line
sns.scatterplot(data=df, x='Study_Hours', y='Exam_Score', hue='Passed', palette='coolwarm', ax=axes[0])
axes[0].set_title('Study Hours vs. Exam Score')

# Correlation Heatmap
numeric_cols = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_cols.corr(), annot=True, cmap='Blues', fmt='.2f', ax=axes[1])
axes[1].set_title('Correlation Heatmap')

plt.tight_layout()
plt.show()
```
