```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
print('OpenCV Version:', cv2.__version__)
```

## 1. Creating and Inspecting a Synthetic Test Image
We create a 300x300 image with geometric shapes (circle, rectangle) to test CV algorithms.

```python
img = np.zeros((300, 300, 3), dtype=np.uint8)
# Draw rectangle (Blue in BGR: (255, 0, 0))
cv2.rectangle(img, (30, 30), (120, 150), (255, 0, 0), -1)
# Draw circle (Green in BGR: (0, 255, 0))
cv2.circle(img, (200, 180), 50, (0, 255, 0), -1)

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Synthetic Test Scene')
plt.axis('off')
plt.show()
```

## 2. Grayscale Conversion & Canny Edge Detection
Canny edge detector identifies pixel intensity discontinuities.

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
edges = cv2.Canny(gray, threshold1=50, threshold2=150)

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].imshow(gray, cmap='gray')
axes[0].set_title('Grayscale Image')
axes[0].axis('off')

axes[1].imshow(edges, cmap='gray')
axes[1].set_title('Canny Edge Map')
axes[1].axis('off')
plt.show()
```

## 3. Finding and Drawing Contours

```python
contours, _ = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
contour_img = img.copy()
for cnt in contours:
    x, y, w, h = cv2.boundingRect(cnt)
    cv2.rectangle(contour_img, (x, y), (x + w, y + h), (0, 0, 255), 2)

plt.figure(figsize=(5, 5))
plt.imshow(cv2.cvtColor(contour_img, cv2.COLOR_BGR2RGB))
plt.title(f'Detected {len(contours)} Objects with Bounding Boxes')
plt.axis('off')
plt.show()
```
